````python
import os
import urllib.parse
import librosa
import librosa.display
import soundfile as sf
import noisereduce as nr
import numpy as np
import matplotlib
matplotlib.use('Agg')  # headless backend, required in Colab for saving without a display
import matplotlib.pyplot as plt
from birdnetlib import Recording
from birdnetlib.analyzer import Analyzer
from transformers import pipeline

USER_LATITUDE = None
USER_LONGITUDE = None

print("🧠 Initializing BirdNET Deep Learning Engine...")
birdnet_model = Analyzer()

print("🤖 Synchronizing Local Creative LLM Prompt Pipelines...")
llm_generator = pipeline('text-generation', model='gpt2', max_new_tokens=30, pad_token_id=50256)

print("✅ Intelligent Multimodal Infrastructure Fully Online.")

def set_coordinates(lat, lon):
    global USER_LATITUDE, USER_LONGITUDE
    USER_LATITUDE = lat
    USER_LONGITUDE = lon
    print(f"\n🌍 Geolocation Filter Synced: Latitude={USER_LATITUDE}, Longitude={USER_LONGITUDE}")

output.register_callback('notebook.SetCoordinates', set_coordinates)

def clean_audio_noise(input_path, output_path):
    try:
        y, sr = librosa.load(input_path, sr=None)
        reduced_noise_audio = nr.reduce_noise(y=y, sr=sr, prop_decrease=0.85)
        sf.write(output_path, reduced_noise_audio, sr)
        return True
    except Exception:
        return False

# ============================================================
# 🔥 NEW: Frequency spectrogram generator
# ============================================================
def generate_frequency_plot(audio_path):
    """
    Loads the audio clip, computes its STFT, and renders a frequency
    spectrogram (time vs. frequency, colored by magnitude in dB).
    Returns a base64 PNG data URI ready to drop straight into an <img src="...">.
    Returns "" if plotting fails, so a plot failure never breaks detection.
    """
    try:
        y, sr = librosa.load(audio_path, sr=None)
        D = librosa.amplitude_to_db(np.abs(librosa.stft(y)), ref=np.max)

        fig, ax = plt.subplots(figsize=(4.2, 2.3), dpi=100)
        librosa.display.specshow(D, sr=sr, x_axis='time', y_axis='hz', ax=ax, cmap='magma')
        ax.set_title("Frequency Spectrogram", fontsize=9)
        ax.set_xlabel("Time (s)", fontsize=8)
        ax.set_ylabel("Frequency (Hz)", fontsize=8)
        ax.tick_params(labelsize=7)
        fig.tight_layout()

        buf = io.BytesIO()
        fig.savefig(buf, format='png', bbox_inches='tight')
        plt.close(fig)  # free memory — important since this runs in a loop
        buf.seek(0)

        encoded = base64.b64encode(buf.read()).decode('utf-8')
        return f"data:image/png;base64,{encoded}"
    except Exception as e:
        print(f"⚠️ Frequency plot generation failed: {e}")
        return ""

def generate_llm_translation(bird_name):
    prompt = f"The {bird_name} chirps enthusiastically and says: \""
    try:
        res = llm_generator(prompt, num_return_sequences=1)
        full_text = res[0]['generated_text']
        quote_start = full_text.find('"')
        if quote_start != -1:
            quote_end = full_text.find('"', quote_start + 1)
            if quote_end != -1:
                return f"✨ {full_text[quote_start:quote_end+1]}"
            return f"✨ {full_text[quote_start:]}\""
        return f"✨ \"Hello human! Singing loud and clear from our shared ecosystem!\""
    except Exception:
        return "✨ 'Chirping away happily in the sun!'"

def generate_bird_image_url(bird_name):
    clean_name = urllib.parse.quote(bird_name.split('(')[0].strip())
    return f"https://unsplash.com{clean_name}"

# ============================================================
# 🔥 UPDATED: generates the frequency plot for the clip and forwards it to the UI
# ============================================================
def analyze_and_translate(file_path):
    global USER_LATITUDE, USER_LONGITUDE
    try:
        if USER_LATITUDE is not None and USER_LONGITUDE is not None:
            recording = Recording(
                birdnet_model,
                file_path,
                lat=USER_LATITUDE,
                lon=USER_LONGITUDE,
                min_conf=0.35
            )
        else:
            recording = Recording(birdnet_model, file_path, min_conf=0.35)

        recording.analyze()

        if recording.detections:
            # Generate the spectrogram once per clip — reused for every detection in it
            freq_plot_url = generate_frequency_plot(file_path)

            for detection in recording.detections:
                name = detection['common_name']
                conf = detection['confidence']
                dynamic_translation = generate_llm_translation(name)
                dynamic_image_url = generate_bird_image_url(name)
                sender_label = f"{name} ({conf*100:.0f}% Match)"

                push_message_to_ui(
                    sender_label,
                    dynamic_translation,
                    is_bird=True,
                    image_url=dynamic_image_url,
                    bird_common_name=name,
                    freq_plot_url=freq_plot_url
                )
        else:
            push_message_to_ui("System Logger", "Processed 3s audio frame: Static ambient noise suppressed.", is_bird=False, image_url="", bird_common_name="")
    except Exception as e:
        push_message_to_ui("System Error", f"Inference breakdown: {str(e)}", is_bird=False, image_url="", bird_common_name="")
````

````python
RAW_WAV_PATH = "raw_segment.wav"
CLEAN_WAV_PATH = "cleaned_segment.wav"

def process_audio_chunk(base64_audio_string):
    header, encoded = base64_audio_string.split(",", 1)
    audio_bytes = base64.b64decode(encoded)
    with open(RAW_WAV_PATH, "wb") as f:
        f.write(audio_bytes)

    filter_success = clean_audio_noise(RAW_WAV_PATH, CLEAN_WAV_PATH)
    target_path = CLEAN_WAV_PATH if filter_success else RAW_WAV_PATH
    analyze_and_translate(target_path)

output.register_callback('notebook.ProcessAudioChunk', process_audio_chunk)
````

````python
import io
import os
import urllib.parse
import librosa
import soundfile as sf
import noisereduce as nr
from birdnetlib import Recording
# 🔥 FIXED IMPORT: Changed from birdnetlib.models to birdnetlib.analyzer
from birdnetlib.analyzer import Analyzer
from transformers import pipeline

# Global variables to store real-time user location metrics
USER_LATITUDE = None
USER_LONGITUDE = None

# 1. Initialize Neural Acoustic Architecture
print("🧠 Initializing BirdNET Deep Learning Engine...")
# 🔥 FIXED INITIALIZATION: Using Analyzer() instead of BirdNETModel()
birdnet_model = Analyzer()

print("🤖 Synchronizing Local Creative LLM Prompt Pipelines...")
llm_generator = pipeline('text-generation', model='gpt2', max_new_tokens=30, pad_token_id=50256)
print("✅ Intelligent Multimodal Infrastructure Fully Online.")

def set_coordinates(lat, lon):
    """Callback linked to browser JS to inject coordinates into the backend engine."""
    global USER_LATITUDE, USER_LONGITUDE
    USER_LATITUDE = lat
    USER_LONGITUDE = lon
    print(f"\n🌍 Geolocation Filter Synced: Latitude={USER_LATITUDE}, Longitude={USER_LONGITUDE}")

output.register_callback('notebook.SetCoordinates', set_coordinates)

def clean_audio_noise(input_path, output_path):
    try:
        y, sr = librosa.load(input_path, sr=None)
        reduced_noise_audio = nr.reduce_noise(y=y, sr=sr, prop_decrease=0.85)
        sf.write(output_path, reduced_noise_audio, sr)
        return True
    except Exception:
        return False

def generate_llm_translation(bird_name):
    prompt = f"The {bird_name} chirps enthusiastically and says: \""
    try:
        res = llm_generator(prompt, num_return_sequences=1)
        full_text = res['generated_text']
        quote_start = full_text.find('"')
        if quote_start != -1:
            quote_end = full_text.find('"', quote_start + 1)
            if quote_end != -1:
                return f"✨ {full_text[quote_start:quote_end+1]}"
            return f"✨ {full_text[quote_start:]}\""
        return f"✨ \"Hello human! Singing loud and clear from our shared ecosystem!\""
    except Exception:
        return "✨ 'Chirping away happily in the sun!'"

def generate_bird_image_url(bird_name):
    clean_name = urllib.parse.quote(bird_name.split('(')[0].strip())
    return f"https://unsplash.com{clean_name}"

def analyze_and_translate(file_path):
    global USER_LATITUDE, USER_LONGITUDE
    try:
        # Check if live browser geolocation filters are active, and append coordinates dynamically
        if USER_LATITUDE is not None and USER_LONGITUDE is not None:
            recording = Recording(
                birdnet_model,
                file_path,
                lat=USER_LATITUDE,
                lon=USER_LONGITUDE,
                min_conf=0.35
            )
        else:
            recording = Recording(birdnet_model, file_path, min_conf=0.35)

        recording.analyze()

        if recording.detections:
            for detection in recording.detections:
                name = detection['common_name']
                conf = detection['confidence']

                dynamic_translation = generate_llm_translation(name)
                dynamic_image_url = generate_bird_image_url(name)

                sender_label = f"{name} ({conf*100:.0f}% Match)"
                # Pass 'name' down to trigger the statistics counter increments
                push_message_to_ui(sender_label, dynamic_translation, is_bird=True, image_url=dynamic_image_url, bird_common_name=name)
        else:
            push_message_to_ui("System Logger", "Processed 3s audio frame: Static ambient noise suppressed.", is_bird=False, image_url="", bird_common_name="")

    except Exception as e:
        push_message_to_ui("System Error", f"Inference breakdown: {str(e)}", is_bird=False, image_url="", bird_common_name="")
````

````python
import io
import os
import time
import datetime
import urllib.parse
import numpy as np
import librosa
import soundfile as sf
import noisereduce as nr
from birdnetlib import Recording
from birdnetlib.analyzer import Analyzer
from transformers import pipeline

print("🧠 Loading Unified Acoustic Processing Engines...")
birdnet_model = Analyzer()

# Load a larger text generation pipeline for more complex and grammatically diverse output
llm_generator = pipeline('text-generation', model='gpt2-medium', max_new_tokens=60, pad_token_id=50256)
print("✅ Advanced Bioacoustic Model Stack Online.")

# Global state trackers for social interaction metrics
LAST_DETECTION_TIME = None
LAST_DETECTED_SPECIES = None

def extract_advanced_syllables(y, sr):
    """
    Bioacoustic Segmentation: Analyzes energy changes in audio to count
    individual vocal notes (syllables) and measure song tempo.
    """
    try:
        # Compute short-time energy
        rmse = librosa.feature.rms(y=y)[0]
        # Find points where energy rises above ambient threshold
        threshold = np.mean(rmse) * 1.2
        syllables = np.where(rmse > threshold)[0]

        # Group adjacent frames into distinct syllable blocks
        if len(syllables) == 0:
            return 0, 0.0

        diff = np.diff(syllables)
        distinct_notes = np.sum(diff > 2) + 1

        # Calculate tempo / speed of vocal delivery
        onset_env = librosa.onset.onset_strength(y=y, sr=sr)
        tempo = librosa.feature.tempo(onset_envelope=onset_env, sr=sr)[0]

        return int(distinct_notes), float(tempo)
    except Exception:
        return 1, 120.0

def decode_avian_semantics(bird_name, note_count, tempo):
    """
    Computational Linguistics Layer: Translates structural properties
    (note count, speed, time of day) into human emotional phrasing.
    """
    now = datetime.datetime.now()
    current_hour = now.hour

    # 1. Determine Time Context
    if 4 <= current_hour <= 8:
        time_context = "during the energetic dawn chorus, establishing daily structural territory"
    elif 11 <= current_hour <= 14:
        time_context = "during midday foraging, focusing primarily on food distribution and tracking resources"
    else:
        time_context = "during evening roosting operations, communicating status to the local group"

    # 2. Determine Structural Urgency via Tempo and Notes
    if tempo > 140:
        urgency = "high-frequency rapid calls, indicating heightened physiological arousal or external alert signals"
        intent = "Attention group! Rapid status change detected in immediate proximity. Check positions!"
    elif note_count > 6:
        urgency = "a complex, multi-syllable sequence with high structural variety"
        intent = "This specific perimeter sector is thoroughly claimed by me. Evaluating available mates and balancing position signals."
    else:
        urgency = "low-density localized contact chirps"
        intent = "Position verified. Safe baseline. Continuing normal ambient operations."

    # 3. Construct an advanced linguistic prompt for the LLM
    prompt = (
        f"A bioacoustic system analyzed a {bird_name} {time_context}. "
        f"The song structural properties showed {urgency}. "
        f"The calculated linguistic meaning translates precisely to human words as: \""
    )

    try:
        res = llm_generator(prompt, num_return_sequences=1)
        full_text = res[0]['generated_text']

        # Extract the generated text inside the quotes cleanly
        quote_start = full_text.find('"', len(prompt) - 20)
        if quote_start != -1:
            quote_end = full_text.find('"', quote_start + 1)
            if quote_end != -1:
                return f"📖 {full_text[quote_start:quote_end+1]}"
            return f"📖 {full_text[quote_start:]}\""
        return f"📖 \"{intent}\""
    except Exception:
        return f"📖 \"{intent}\""

def generate_bird_image_url(bird_name):
    clean_name = urllib.parse.quote(bird_name.split('(')[0].strip())
    return f"https://unsplash.com{clean_name}"

def analyze_and_translate(file_path):
    global LAST_DETECTION_TIME, LAST_DETECTED_SPECIES
    try:
        # Load audio data array to extract advanced signal features
        y, sr = librosa.load(file_path, sr=None)
        note_count, tempo = extract_advanced_syllables(y, sr)

        recording = Recording(birdnet_model, file_path, min_conf=0.35)
        recording.analyze()

        if recording.detections:
            for detection in recording.detections:
                name = detection['common_name']
                conf = detection['confidence']

                # Check for social correlation: did a bird reply to another bird?
                current_time = time.time()
                social_note = ""
                if LAST_DETECTION_TIME and (current_time - LAST_DETECTION_TIME) < 10:
                    if LAST_DETECTED_SPECIES == name:
                        social_note = "🔄 [Social Note: Direct repetitive phrase confirmation match]\n"
                    else:
                        social_note = f"🤝 [Social Note: Cross-species acoustic counter-response with {LAST_DETECTED_SPECIES}]\n"

                # Update tracking variables
                LAST_DETECTION_TIME = current_time
                LAST_DETECTED_SPECIES = name

                # Run the upgraded deep translation generation layer
                advanced_translation = decode_avian_semantics(name, note_count, tempo)
                final_output = f"{social_note}🔢 Notes: {note_count} | Speed: {int(tempo)} BPM\n{advanced_translation}"

                dynamic_image_url = generate_bird_image_url(name)
                sender_label = f"{name} ({conf*100:.0f}% Semantics Match)"

                push_message_to_ui(sender_label, final_output, is_bird=True, image_url=dynamic_image_url, bird_common_name=name)
        else:
            push_message_to_ui("System Logger", "Processed 3s audio window: Isolate audio filters stabilizing...", is_bird=False, image_url="", bird_common_name="")

    except Exception as e:
        push_message_to_ui("System Error", f"Advanced Interpretation Error: {str(e)}", is_bird=False, image_url="", bird_common_name="")
````

````python
RAW_WAV_PATH = "raw_segment.wav"
CLEAN_WAV_PATH = "cleaned_segment.wav"

def process_audio_chunk(base64_audio_string):
    header, encoded = base64_audio_string.split(",", 1)
    audio_bytes = base64.b64decode(encoded)

    with open(RAW_WAV_PATH, "wb") as f:
        f.write(audio_bytes)

    filter_success = clean_audio_noise(RAW_WAV_PATH, CLEAN_WAV_PATH)
    target_path = CLEAN_WAV_PATH if filter_success else RAW_WAV_PATH

    analyze_and_translate(target_path)

output.register_callback('notebook.ProcessAudioChunk', process_audio_chunk)

````
